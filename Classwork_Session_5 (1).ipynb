```python
def my_factorial(number):
    if number < 0:
        return None
    if number == 0:
        return 1
    
    result = 1
    print(result, end = '')
    for i in range (2, number + 1):
        result *= i
        print(" *", i, end = "")

    print(' = ', result)    
    return result

print(my_factorial(5))
print(my_factorial(-5))
print(my_factorial(0))
```

```python
def my_fib(n):
    if n < 1:
        return None
    if n==1 or n==2:
        return 1

    f1 = 1
    f2 = 1
    fn = 0

    for i in range (3, n+1):
        fn = f1 + f2
        f1 = f2
        f2 = fn
        
    return fn

print(my_fib(12))
```
