```python
text = "привет это секрет не читай плз!"
secret = 55

secret_message = ""

for char in text:
    secret_message += chr(ord(char) + secret)

print(text)
print(secret_message)

print()
print("расшифрованное сообщение:")
for char in secret_message:
    print(chr(ord(char)-secret), end="")
```

```python
text = "привет это секрет не читай плз!"
secret = 55

secret_message = "LOLOLOLOL"

for char in text:
    code_for_char = ord(char)
    new_code_for_char = code_for_char + secret

secret_message += chr(new_code_for_char)

print(text)
print(secret_message)

print()
print("расшифрованное сообщение:")

for char in secret_message:
    code_for_char = ord(char)
    new_code_for_char = code_for_char - secret

print(chr(new_code_for_char), end="")
```
