```python
def my_fib(n):
    if n < 1:
        return None
    if n==1 or n==2:
        return 1

    f1 = 1
    f2 = 1
    fn = 0

    for i in range (3, n+1):
        fn = f1 + f2
        f1 = f2
        f2 = fn
        
    return fn

print(my_fib(12))
```

```python
import calendar


year = int(input("Enter your year that you would like to check:"))
if calendar.isleap(year) == True:
    print(f"The year {year} is a leap year")
else:
    print(f"The year {year} is not a leap year")
```

```python
weight = int(input("Input your weight in kilograms:"))
height = int(input("Input your height in meters:"))

BMI = weight / (height ** 2)
BMI = round(BMI, 3)

if BMI < 18.5:
    print(f"With the height of {height} and weight of {weight}, your BMI is {BMI} which is considered under the norm")
if 18.5 < BMI < 24.9:
    print(f"With the height of {height} and weight of {weight}, your BMI is {BMI} which is considered normal")
if 25.0 < BMI:
    print(f"With the height of {height} and weight of {weight}, your BMI is {BMI} which is considered over the norm")
```
